# Telco Churn EDA

Exploratory data analysis on the synthetic Telco Customer Churn dataset.
Matches the schema and distributions from the public Telco Churn dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('ml/data/raw/telco_churn.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
print('Churn distribution:')
print(df['churn'].value_counts())
print(f'Churn rate: {(df["churn"] == "Yes").mean():.1%}')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
sns.countplot(data=df, x='churn', ax=axes[0, 0])
axes[0, 0].set_title('Churn Distribution')

sns.boxplot(data=df, x='churn', y='tenure', ax=axes[0, 1])
axes[0, 1].set_title('Tenure by Churn')

sns.boxplot(data=df, x='churn', y='monthly_charges', ax=axes[1, 0])
axes[1, 0].set_title('Monthly Charges by Churn')

sns.countplot(data=df, x='contract', hue='churn', ax=axes[1, 1])
axes[1, 1].set_title('Contract Type vs Churn')

plt.tight_layout()
plt.show()
```

```python
categorical = ['gender', 'partner', 'dependents', 'phone_service',
               'multiple_lines', 'internet_service', 'online_security',
               'online_backup', 'device_protection', 'tech_support',
               'streaming_tv', 'streaming_movies', 'contract',
               'paperless_billing', 'payment_method']

fig, axes = plt.subplots(4, 4, figsize=(16, 14))
axes = axes.flatten()

for i, col in enumerate(categorical):
    churn_rate = df.groupby(col)['churn'].apply(lambda x: (x == 'Yes').mean())
    churn_rate.plot(kind='bar', ax=axes[i])
    axes[i].set_title(f'Churn Rate by {col}')
    axes[i].tick_params(axis='x', rotation=45)

# Hide empty subplots
for j in range(len(categorical), len(axes)):
    axes[j].axis('off')

plt.tight_layout()
plt.show()
```

```python
numeric = ['tenure', 'monthly_charges', 'total_charges']
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, col in enumerate(numeric):
    sns.histplot(data=df, x=col, hue='churn', bins=30, element='step', ax=axes[i])
    axes[i].set_title(f'{col} Distribution by Churn')

plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap for numeric features
df_num = df.select_dtypes(include=[np.number])
plt.figure(figsize=(10, 8))
sns.heatmap(df_num.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Numeric Feature Correlation')
plt.show()
```

## Key Findings

- Churn rate is approximately 25-30% (synthetic, depends on seed).
- Month-to-month contracts have significantly higher churn.
- Electronic check payment method correlates with higher churn.
- Fiber optic internet customers churn more.
- Tenure is negatively correlated with churn.
- Monthly charges show moderate positive correlation with churn.
